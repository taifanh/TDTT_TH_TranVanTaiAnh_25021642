```python
# 15
def find_movies(path):
    movies = []

    with open(path, "r", encoding="utf-8") as file:
        next(file)
        for line in file:
            cols = line.strip().split(",")

            rating = float(cols[3])
            genre = cols[2].lower()

            if rating >= 8.0 and genre != "boring":
                movies.append(cols[1])

    return movies if movies else "All are boring"


print(*find_movies("test.txt"))
```

```python
# 17
def productExceptSelf(path):
    file = open(path, "r")
    n = int(file.readline().strip())
    nums = list(map(int, file.readline().strip().split()))
    file.close()
    ans = []
    for i in range(n):
        prod = 1
        for j in range(n):
            if i != j:
                prod *= nums[j]
        ans.append(prod)
    for x in ans:
        print(x, end=" ")
productExceptSelf("test.txt")
```
