```python
# W6A19
def cal(a):
    for i in range(len(a)):
        if a[i] == max(a):
            return i + 1
    return -1
a = list(map(int, input().split()))
print(cal(a))
```

```python
# W6A20
def check(a, k):
    for i in range(1, len(a)):
        if a[i - 1] == k:
            return i
    return -1
a = list(map(int, input().split()))
k = int(input())
print(check(a, k))
```
