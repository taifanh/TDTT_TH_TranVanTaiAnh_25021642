```python
# W9A4
def solve(a, b):
    pos = [[-1] * 26 for _ in range(len(a))]
    for i in range(len(a)):
        for j in range(26):
            # print(id(pos[6]), end = " ")
            if ord(a[i]) - ord('a') == j:
                pos[i][j] = i
                # print(i, j, pos[i][j], pos[6][1], end = " ")
            elif i > 0:
                pos[i][j] = pos[i - 1][j]
            # print(f"i={i}", f"j={j}", id(pos[i]), id(pos[6]))
    # print(pos[6][1])
    j = len(a) - 1
    for i in range(len(b) - 1, -1, -1):
        if j == -1:
            return "NO"
        # print(j, ord(b[i]) - ord('a'), end = " ")
        j = pos[j][ord(b[i]) - ord('a')]
        # print(j, end = " ")
        if j == -1:
            return "NO"
        # print("|", b[i], j, a[j])
        j -= 1
    return "YES"

q = int(input())
while q > 0:
    a = input().lower()
    b = input().lower()
    print(solve(a, b))
    q -= 1
```

```python
# W9A5
list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))
def solve(list1, list2):
    if len(list1) > len(list2):
        return "NO"
    for i in range(len(list2) - len(list1)):
        if list1 == list2[i : i + len(list1)]:
            return "YES"
    return "NO"
print(solve(list1, list2))
```
