```python
# W9A4
def solve(a, b):
    pos = [[-1] * 26 for _ in range(len(a))]
    for i in range(len(a)):
        for j in range(26):
            # print(id(pos[6]), end = " ")
            if ord(a[i]) - ord('a') == j:
                pos[i][j] = i
                # print(i, j, pos[i][j], pos[6][1], end = " ")
            elif i > 0:
                pos[i][j] = pos[i - 1][j]
            # print(f"i={i}", f"j={j}", id(pos[i]), id(pos[6]))
    # print(pos[6][1])
    j = len(a) - 1
    for i in range(len(b) - 1, -1, -1):
        if j == -1:
            return "NO"
        # print(j, ord(b[i]) - ord('a'), end = " ")
        j = pos[j][ord(b[i]) - ord('a')]
        # print(j, end = " ")
        if j == -1:
            return "NO"
        # print("|", b[i], j, a[j])
        j -= 1
    return "YES"

q = int(input())
while q > 0:
    a = input().lower()
    b = input().lower()
    print(solve(a, b))
    q -= 1
```

```python
# W9A6
def next_permutation(nums):
    n = len(nums)
    i = n - 2
    while i >= 0 and nums[i] >= nums[i + 1]:
        i -= 1
    
    if i >= 0:
        j = n - 1
        while nums[j] <= nums[i]:
            j -= 1
        nums[i], nums[j] = nums[j], nums[i]
    left, right = i + 1, n - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
        
    return nums
n = int(input())
nums = list(map(int, input().split()))
next_nums = next_permutation(nums)
for x in nums:
    print(x, end = " ")
```

```python
# W9A7
def canPlaceFlowers(flowerbed, k):
    if k == 0:
        return True
        
    n = len(flowerbed)
    
    for i in range(n):
        if flowerbed[i] == 0:
            l = (i == 0) or (flowerbed[i - 1] == 0)
            r = (i == n - 1) or (flowerbed[i + 1] == 0)
            
            if l and r:
                flowerbed[i] = 1
                k -= 1
                
                if k == 0:
                    return True
    return False
a = list(map(int, input().split()))
k = int(input())
print(canPlaceFlowers(a, k))
```

```python
# W9A8
def is_path_crossing(moves):
    x, y = 0, 0
    visited = {(0, 0)}
    
    for move in moves:
        if move == 'U':
            y += 1
        elif move == 'D':
            y -= 1
        elif move == 'R':
            x += 1
        elif move == 'L':
            x -= 1
            
        if (x, y) in visited:
            return True
        visited.add((x, y))
        
    return False
moves = input()
print(is_path_crossing(moves))
```

```python
# W9A9
def Median(nums):
    if nums[0] > nums[1]:
        nums[0], nums[1] = nums[1], nums[0]

    if nums[2] > nums[3]:
        nums[2], nums[3] = nums[3], nums[2]

    if nums[0] > nums[2]:
        nums[0], nums[2] = nums[2], nums[0]
        nums[1], nums[3] = nums[3], nums[1]

    if nums[1] > nums[4]:
        nums[1], nums[4] = nums[4], nums[1]

    if nums[1] > nums[2]:
        nums[1], nums[2] = nums[2], nums[1]
        nums[4], nums[3] = nums[3], nums[4]

    if nums[2] > nums[4]:
        nums[2], nums[4] = nums[4], nums[2]
        
    return nums[2]
nums = list(map(int, input().split()))
print(Median(nums))
```

```python
# W9A10
import bisect
def findRadius(houses, heaters):
    heaters.sort()
    radius = 0
    for h in houses:
        i = bisect.bisect_right(heaters, h)
        dist_left = h - heaters[i-1] if i > 0 else float('inf')
        dist_right = heaters[i] - h if i < len(heaters) else float('inf')
        min_dist = min(dist_left, dist_right)
        radius = max(radius, min_dist)
        
    return radius
houses = list(map(int, input().split()))
heaters = list(map(int, input().split()))
print(findRadius(houses, heaters))
```
