```python
# W10A1
try:
    a, b = map(int, input().split())
    print(f"{a / b:.2f}")
except ZeroDivisionError:
    print("Loi chia khong")
```

```python
# W10A2
n = list(map(int, input().split()))
try:
    print(n[len(n)])
except IndexError:
    print("Truy cap sai chi so")
```

```python
# W10A3
def bp(a):
    sum = 0
    while a > 0:
        sum += (a % 10) ** 2
        a //= 10
    return sum
try:
    a = int(input())
    check = []
    while a not in check and a != 1:
        check.append(a)
        a = bp(a)
    if a != 1:
        print("NO")
    else:
        print("YES")
except ValueError:
    print("Nhap so")
```

```python
# W10A4
try:
    a = float(input("Nhập số thứ nhất: "))
    b = float(input("Nhập số thứ hai: "))
    
except ValueError:
    print("Lỗi: Bạn vui lòng chỉ nhập số!")
    
else:
    tong = a + b
    print(f"Tổng hai số là: {tong}")
    
finally:
    print("Kết thúc chương trình.")
```
