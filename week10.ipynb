```python
try:
    a, b = map(int, input().split())
    print(f"{a / b:.2f}")
except ZeroDivisionError:
    print("Loi chia khong")
```

```python
n = list(map(int, input().split()))
try:
    print(n[len(n)])
except IndexError:
    print("Truy cap sai chi so")
```

```python
def bp(a):
    sum = 0
    while a > 0:
        sum += (a % 10) ** 2
        a //= 10
    return sum
try:
    a = int(input())
    check = []
    while a not in check and a != 1:
        check.append(a)
        a = bp(a)
    if a != 1:
        print("NO")
    else:
        print("YES")
except ValueError:
    print("Nhap so")
```
